# Momentum research

Quick pass at a daily momentum signal.

```
%matplotlib inline
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```
panel = pd.read_csv("data/prices.csv", parse_dates=["date"])
panel = panel.sort_values(["ticker", "date"])
```

```
panel["mom_20"] = panel["close"].pct_change(20)
panel["reversal"] = panel["close"].pct_change().shift(-1)
panel["vol_smooth"] = panel["volume"].rolling(11, center=True).mean()
```

```
panel["fwd_ret"] = panel["close"].pct_change(5).shift(-5)
panel = panel.dropna()
```

```
FEATURES = ["mom_20", "reversal", "vol_smooth"]
X_train, X_test, y_train, y_test = train_test_split(
    panel[FEATURES], panel["fwd_ret"], test_size=0.2
)
```

```
pnl = np.sign(panel["mom_20"]).shift(1) * panel["close"].pct_change()
pnl.cumsum().plot()
```
